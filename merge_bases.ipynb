# Por aqui começaremos a fazer o merge das tabelas para depois começarmos o EDA profundo

```python
import pandas as pd
# Carrega o arquivo que você arrumou ontem
df_logistica = pd.read_csv('../data/logistica.csv')

# Carrega as outras tabelas do projeto 
df_pedidos = pd.read_csv('../data/pedidos.csv')
df_clientes = pd.read_csv('../data/clientes.csv')
df_itens = pd.read_csv('../data/itens_vendas.csv')
```

```python
# Intens vendas é o coraçao do projeto
linhas_iniciais = df_itens.shape[0]
print(f"Linhas iniciais: {linhas_iniciais}")
```

```python
# 1. Cruzamento inicial: Itens (coração) + Pedidos (capa do pedido)
# Estamos assumindo que a coluna que liga as duas se chama 'id_pedido'. 
# (Ajuste se o nome da coluna for diferente)
df_merge_1 = pd.merge(df_itens, df_pedidos, on='ID_Pedido', how='left')

# prova real imediata
linhas_apos_merge = df_merge_1.shape[0]
print(f"Linhas após juntar itens e pedidos: {linhas_apos_merge}")


# vendo se juntou as tabelas
display(df_merge_1.head())
```

# 3. Cruzamento: Base atual (Itens + Pedidos) + Clientes
# A chave que liga as duas tabelas agora é o 'ID_Cliente'


```python
df_merge_2 = pd.merge(df_merge_1, df_clientes, on='ID_Cliente', how='left')
# Prova real
linhas_com_clientes =df_merge_2.shape[0]
print(f"Linhas após juntar clientes: {linhas_com_clientes}")

# Vendo o resultado (deve aparecer nome, estado, etc. lá no final das colunas)
display(df_merge_2.head())
```

```python
# 4. Último Merge: Base atual (Itens + Pedidos + Clientes) + Logística
# Ajuste o 'on' se a coluna chave na tabela de logística tiver um nome diferente (ex: id_pedido minúsculo)
df_final = pd.merge(df_merge_2, df_logistica, on='ID_Pedido', how='left')

# Checagem de segurança (tem que manter as 11 linhas iniciais)
print(f"Linhas do Dataset Master: {df_final.shape[0]}")

# Salva o arquivo final consolidado para usar no próximo passo
df_final.to_csv('../data/dataset_consolidado.csv', index=False)

display(df_final.head())
```
